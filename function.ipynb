##1.What is the difference between a function and a method in Python ?
==>> **Function** :A function is a reusable block of code that performs a specific task.
It is defined using the def keyword (or lambda for anonymous functions) and can be called independently of any object.


```python
def func(x):
  return x**2
print(func(5))
```

It is called using dot notation (object.method()) and automatically takes the object as the first argument (usually named self in instance methods).

```python
my_list=[2, 3, 1, 9, 5, 7]
my_list.sort()
print(my_list)
```

##7.What are the advantages of using generators over regular functions?
==>>Advantages of Generators over Regular Functions:

**Memory Efficient** – Generate items one at a time without storing the entire sequence in memory.

**Lazy Evaluation**– Values are produced only when needed, speeding up initial azy Evaluationexecution.

**Handle Infinite Sequences** – Can model endless data streams without memory issues.

**Cleaner Code** – Easier to write than full iterator classes.

**Better Performance** – Faster for large data when full list is not required.

**Pipeline Processing** – Can chain generators to process data in steps efficiently.

```python
def gen():
    for i in range(1000000):
        yield i
#avoids storing a list of 1,000,000 numbers in memory.
```

```python
g = gen()
```

```python
next(g)
```

```python
next(g) #and so on
```

##10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
==>>map() applies a given function to each item in an iterable and returns a new iterable with the transformed values.

```python
nums = [1, 2, 3, 4]
print(list(map(lambda x: x**2, nums)))
```

filter() applies a function to each item and returns a new iterable containing only the items that return True for the function.

```python
print(list(filter(lambda x: x % 2 == 0, nums)))
```

reduce() (from the functools module) applies a function cumulatively to the items, reducing the iterable to a single result.


```python
from functools import reduce
print(reduce(lambda x, y: x * y, nums))
```

##4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(number):
    if number < 1 or number > 200:
        return False
    if number == 1:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
```

```python
is_prime(8)
```

```python
is_prime(3)
```

```python
is_prime(199)
```

```python
is_prime(197)
```

```python
is_prime(193)
```

```python
is_prime(55)
```

##5.Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.



```python
def fibonacci_generator(max_terms):
    a, b = 0, 1
    for _ in range(max_terms):
        yield a
        a, b = b, a + b

for num in fibonacci_generator(8):
    print(num, end=" ")
```

##6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp

for value in powers_of_two(5):
    print(value)

```

```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.rstrip('\n')
```

##.10 Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def remove_vowels(input_str):
    vowels = 'aeiouAEIOU'
    result = ''.join(filter(lambda ch: ch not in vowels, input_str))
    return result

text = "Hello, how are you?"
no_vowels = remove_vowels(text)

print("Original:", text)
print("Without vowels:", no_vowels)
```
